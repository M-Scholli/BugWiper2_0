```python
# load the data

from pathlib import Path
import os
import re

from pathlib import Path
import pandas as pd

file_path = Path(
    "../prototype_1",
    "flight_2026_05_15",
    "log_0003.csv"
)


def title_to_filename(title):
    import re
    name = title.lower()
    name = name.replace(" ", "_")
    name = re.sub(r"[^a-z0-9_]", "", name)
    return name


df = pd.read_csv(file_path)

# print(df.head())
```

```python
# convert the Data
df["time_s"] = df["time"] / 1e3
df['position_m'] = df['position'] / 1e3
df['current_A'] = df['motor current in mA'] / 1e3

# df["time_s"] = df["time_s"] - df["time_s"].iloc[0]
```

```python
def save_plot(fig, title, file_path):
    # convert title to filename-safe string
    filename = title_to_filename(title)

    # mirror folder structure
    plot_base = Path("..", "plots") / file_path.relative_to("..")
    plot_base = plot_base.with_suffix("")

    # build full file path
    plot_file = plot_base.parent / f"{plot_base.stem}_{filename}.png"

    # create folder if needed
    os.makedirs(plot_file.parent, exist_ok=True)

    # save figure
    fig.savefig(plot_file, dpi=300, bbox_inches="tight")
```

```python
import matplotlib.pyplot as plt

df["time_s_plot0"] = df["time_s"]

fig, axs = plt.subplots(3, 1, sharex=True)

axs[0].plot(df["time_s_plot0"], df["position_m"])
axs[0].set_ylabel("Position [m]")
axs[0].grid()

axs[1].plot(df["time_s_plot0"], df["speed"])
axs[1].set_ylabel("Speed")
axs[1].grid()

axs[2].plot(df["time_s_plot0"], df["current_A"])
axs[2].set_ylabel("Current [A]")
axs[2].set_xlabel("Time [s]")
axs[2].grid()

#axs[2].set_xlim(0, 35)

plt.suptitle("Flight Log Overview")
save_plot(fig, "Flight Overview", file_path)
plt.show()
```
